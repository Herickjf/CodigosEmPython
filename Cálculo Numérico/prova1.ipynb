##**Questão 1**

**Enunciado:** a velocidade de ascensão de um foguete em voo vertical próximo à superfície terrestre ($v$) pode ser aproximada pela seguinte expressão:
\
$$
v = u \cdot \ln(\frac{M_{0}}{M_{0} - c \cdot t}) + g \cdot t
$$
\
Na qual:
$u$ – velocidade de exaustão relativa ao foguete;
$M_{0}$ – massa do foguete ao ser lançado;
$c$ – taxa de consumo de combustível;
$g$ – Aceleração gravitacional; e
$t$ – tempo (medido a partir do lançamento).
\
\
Considerando $u = 200$ m/s, $M_{0} = 1600$ kg, $g = 9.8$ m/s e $c = 27$ kg/s, determinar o instante em que $v = 100$ m/s. Empregar o método da bissecção para determinar uma raiz aproximada para a equação, com intervalo inicial $[6, 8]$ e tolerância de $0,8\%$. Explicite o erro relativo associado a cada iteração.

```python
# Definindo os simbolos e as expressões

u, M, c, x, g = sp.symbols("u, M0:1, c, t, g")
vl = u * sp.ln(M / (M - c * x)) + g * x; vl
```

```python
# Substituindo os valores especificos das variaveis

v = vl.subs(u, 200).subs(M, 1600).subs(c, 27).subs(g, 9.8); v
```

```python
# Valores iniciais do intervalo: [a, b]
a =  8.0
b =  6.0

# O erro vai ser usado como criterio de parada das iteracoes
erro =  0.08

# Valor ao qual se quer chegar
raiz = 100
```

```python
bisseccao(a, b, erro, raiz, v)
```

##**Questão 2**

**Enunciado:** Um cabo telefônico suspenso entre dois postes tem um peso de
quilogramas-força por metro linear. Considerando que a tensão $T$ na
metade do cabo é obtida a partir resolução da seguinte equação:

\
$$
S = \frac{2 \cdot T}{\alpha} \cdot \sinh(\frac{\alpha \cdot L}{2 \cdot T})
$$
\
Na qual: $S$ é o comprimento do cabo e $L$ é a distância entre os postes.

Utilizar o método da bissecção para determinar a tensão $T$ a partir das
seguintes condições: $S = 32$ m, $L = 30$ m, $\alpha = 0.10$ Kgf. Considerar o
intervalo inicial $[2,3]$ e a tolerância superior ou igual a $1\%$.

```python
#Definindo símbolos e expressões:
x, alpha, L, S = sp.symbols("T, a , L, S")
tensao = (2 * x) / alpha * sp.sinh((alpha * L) / (2 * x)) - S
```

```python
# Substituindo os valores especificos das variaveis

ten = tensao.subs(S, 32).subs(L, 30).subs(alpha, 0.10); ten
```

```python
# Valores iniciais do intervalo: [a, b]
a =  2.0
b =  3.0

# O erro vai ser usado como criterio de parada das iteracoes
erro =  0.01

# Valor ao qual se quer chegar
raiz = 0
```

```python
bisseccao(a, b, erro, raiz, ten)
```
