##Método de Euler

```python
def euler(expr, n, h, x0, y0):
  if(n == 1):
    yn = y0 + h * expr.subs(sp.Symbol('x'), x0).subs(sp.Symbol('y'), y0)
    return yn

  yk = euler(expr, n-1, h, x0, y0)
  yn = yk + h * expr.subs(sp.Symbol('x'), x0+h*(n-1) + h*(n-1)).subs(sp.Symbol('y'), yk)
  return yn
```

##Método de Runge-Kutta

###3º ordem

```python
def rungeKutta3(expr, n, h, x0, y0):
  if(n == 1):
    yn = y0 + 1/9 * rk3Vals(expr, h, x0, y0)
    return yn

  yk = rungeKutta3(expr, n-1, h, x0, y0)
  kVals = rk3Vals(expr, h, x0 + h*(n-1), yk)
  yn = yk + 1/9 * rk3Vals
  return yn

def rk3Vals(expr, h, x, y):
  k1 = h * expr.subs(sp.Symbol('x'), x).subs(sp.Symbol('y'), y)
  k2 = h * expr.subs(sp.Symbol('x'), x + h/2).subs(sp.Symbol('y'), y + k1/2)
  k3 = h * expr.subs(sp.Symbol('x'), x + 3*h/4).subs(sp.Symbol('y'), y + 3*k2/4)

  return (2*k1 + 3*k2 + 4*k3)
```

###4º Ordem

```python
def rungeKutta4(expr, n, h, x0, y0):
  if(n == 1):
    yn = y0 +  1/6 * rk4Vals(expr, h, x0, y0)
    return yn

  yk = rungeKutta4(expr, n-1, h, x0, y0)
  kVals = rk4Vals(expr, h, x0 + h*(n-1), yk)
  yn = yk + 1/6 * kVals

  return yn

def rk4Vals(expr, h, x, y):
  k1 = h * expr.subs(sp.Symbol('x'), x).subs(sp.Symbol('y'), y)
  k2 = h * expr.subs(sp.Symbol('x'), x + h/2).subs(sp.Symbol('y'), y + k1/2)
  k3 = h * expr.subs(sp.Symbol('x'), x + h/2).subs(sp.Symbol('y'), y + k2/2)
  k4 = h * expr.subs(sp.Symbol('x'), x + h).subs(sp.Symbol('y'), y + k3)

  return (k1 + 2*k2 + 2*k3 + k4)
```
